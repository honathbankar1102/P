```python

# DISEASE EXPERT SYSTEM

Rule = {
    "flu": ["fever", "cough", "body pain", "headache", "fatigue"],
    "viral_infection": ["fever", "headache", "fatigue", "body pain", "weakness"],
    "common_cold": ["cold", "sneezing", "runny nose", "sore throat", "cough"],
    "malaria": ["fever", "chills", "sweating", "headache", "body pain"],
    "dengue": ["high fever", "headache", "body pain", "rash", "fatigue"]
}

print("================================")
print("     DISEASE EXPERT SYSTEM")
print("================================")

print("\nAvailable Symptoms:")

all_symptoms = set()

for symptoms in Rule.values():
    all_symptoms.update(symptoms)

for symptom in sorted(all_symptoms):
    print("-", symptom)

# Input symptoms
n = int(input("\nEnter number of symptoms: "))

F = set()

for i in range(n):
    symptom = input("Enter symptom: ").lower().strip()
    F.add(symptom)

# FORWARD CHAINING
def forward_chaining(Rule, F):
    detected = []

    for disease, symptoms in Rule.items():
        matched = set(symptoms) & F
        ratio = len(matched) / len(symptoms)

        if ratio >= 0.60:
            detected.append(disease)

    return detected

# BACKWARD CHAINING
def backward_chaining(Rule, F, goal):
    if goal not in Rule:
        return False

    symptoms = Rule[goal]
    matched = set(symptoms) & F
    ratio = len(matched) / len(symptoms)

    return ratio >= 0.60

# Forward chaining result
detected = forward_chaining(Rule, F)

print("\n========== FORWARD CHAINING ==========")

if detected:
    for disease in detected:
        print("Disease Present:", disease)
else:
    print("No disease detected")

# Backward chaining result
print("\nAvailable Diseases:")

for disease in Rule:
    print("-", disease)

goal = input("\nEnter disease to verify: ").lower().strip()

result = backward_chaining(Rule, F, goal)

print("\n========== BACKWARD CHAINING ==========")

if result:
    print("Disease Present:", goal)
else:
    print("Disease Not Present:", goal)
```
