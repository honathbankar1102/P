```python

occupancy = int(input("Is room occupied? (1/0): "))

while occupancy == 1:

    # Check natural light
    light = int(input("Is there sufficient natural light? (1/0): "))

    if light == 0:
        print("Turning ON Lights")

    # Check room temperature
    temp = int(input("Enter room temperature: "))

    if temp >= 32:
        print("Turning OFF fans")
        print("Turning ON AC at 24 C")

    elif temp < 22:
        print("Turning OFF fans and AC")
        print("Turning ON Heater")

    else:
        print("Turning OFF AC and Heater")
        print("Turning ON Fans")

    # Check air quality
    aqi = int(input("Enter room AQI: "))

    if aqi >= 65:
        print("Turning ON Air Purifier")

    elif aqi <= 45:
        print("Turning OFF Air Purifier")

    else:
        print("Air Purifier status unchanged")

    # Sleep mode
    sleep = input("Do you want sleep mode? (yes/no): ")

    if sleep.lower() == "yes":
        print("Turning OFF Lights")

        if temp >= 26:
            print("Turning OFF Fans")
            print("Turning ON AC at 24 C with 40% power")

        elif temp < 22:
            print("Turning OFF Fans")
            print("Turning ON Heater with 40% power")

        else:
            print("Turning ON Fans")

    # Check occupancy again
    occupancy = int(input("Is room occupied? (1/0): "))

print("Turning OFF System")
print("Good Day!")
```
