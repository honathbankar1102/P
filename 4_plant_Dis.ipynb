```python

# PLANT DISEASE EXPERT SYSTEM

# Initial facts
symptoms = set()

# Knowledge Base (Rules)
rules = [
    ({"powdery_patches", "brown_spots"}, "fungus"),
    ({"yellow_rings", "soft_tissues"}, "bacterial"),
    ({"fungus"}, "use_protectants"),
    ({"bacterial"}, "use_bactericides")
]

# Available symptoms
symptom_options = {
    "1": "powdery_patches",
    "2": "brown_spots",
    "3": "yellow_rings",
    "4": "soft_tissues"
}


# Display symptoms
def display_symptoms():
    print("\nCurrent Symptoms:")
    for s in sorted(symptoms):
        print("-", s)


# Add symptoms
def select_symptoms():
    while True:
        print("\n1. Powdery Patches")
        print("2. Brown Spots")
        print("3. Yellow Rings")
        print("4. Soft Tissues")
        print("5. Continue")

        choice = input("Enter choice: ")

        if choice == "5":
            if symptoms:
                break
            print("Enter at least one symptom.")

        elif choice in symptom_options:
            s = symptom_options[choice]
            symptoms.add(s)
            print(s, "added.")

        else:
            print("Invalid choice.")


# FORWARD CHAINING
def forward_chaining():
    print("\n--- FORWARD CHAINING ---")

    known = set(symptoms)
    print("Initial Facts:", known)

    changed = True

    while changed:
        changed = False

        for conditions, result in rules:
            if conditions.issubset(known) and result not in known:
                known.add(result)
                changed = True
                print("Derived:", result)

    print("\nFinal Facts:", known)

    if "fungus" in known:
        print("Conclusion: Fungal Infection")
        print("Treatment: Use fungal protectants.")

    if "bacterial" in known:
        print("Conclusion: Bacterial Infection")
        print("Treatment: Use bactericides.")

    if "fungus" not in known and "bacterial" not in known:
        print("No known disease rule matched.")


# BACKWARD CHAINING
def backward_prove(goal, known, checked=None):

    if checked is None:
        checked = set()

    if goal in known:
        return True

    if goal in checked:
        return False

    checked = checked | {goal}

    for conditions, result in rules:
        if result == goal:
            if all(backward_prove(c, known, checked)
                   for c in conditions):
                return True

    return False


def backward_chaining():
    print("\n--- BACKWARD CHAINING ---")
    print("1. Fungus")
    print("2. Bacterial")
    print("3. Use Protectants")
    print("4. Use Bactericides")

    goals = {
        "1": "fungus",
        "2": "bacterial",
        "3": "use_protectants",
        "4": "use_bactericides"
    }

    choice = input("Enter goal: ")

    if choice not in goals:
        print("Invalid choice.")
        return

    goal = goals[choice]

    print("\nGoal:", goal)

    if backward_prove(goal, symptoms):
        print("Goal PROVED:", goal)

        if goal == "fungus":
            print("Fungal infection detected.")

        elif goal == "bacterial":
            print("Bacterial infection detected.")

        elif goal == "use_protectants":
            print("Recommendation: Use fungal protectants.")

        elif goal == "use_bactericides":
            print("Recommendation: Use bactericides.")

    else:
        print("Goal NOT PROVED:", goal)


# MAIN PROGRAM
while True:

    symptoms.clear()

    print("\n===== PLANT DISEASE EXPERT SYSTEM =====")

    select_symptoms()

    while True:

        print("\n1. Forward Chaining")
        print("2. Backward Chaining")
        print("3. Re-enter Symptoms")
        print("4. Exit")

        choice = input("Enter choice: ")

        if choice == "1":
            forward_chaining()

        elif choice == "2":
            backward_chaining()

        elif choice == "3":
            break

        elif choice == "4":
            print("Program Ended.")
            exit()

        else:
            print("Invalid choice.")
```
