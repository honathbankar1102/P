```python

# Intelligent Agent Function

def temp_cont(temp):

    print("\n----- Intelligent Agent Analysis -----")
    print(f"Current Furnace Temperature: {temp:.1f}°C")

    # Normal temperature condition
    if 500 <= temp <= 2200:
        print("Temperature is within the normal operating range.")
        print("No heating or cooling required.")
        print("Furnace is operating normally.")

    # High temperature condition
    elif temp > 2200:
        print("WARNING: Temperature exceeding safety range!")
        print("Cooling actuator activated...")

        while temp > 1500:
            temp -= 1

        print("Cooling mechanism completed.")
        print(f"Temperature stabilized at {temp:.1f}°C")

    # Low temperature condition
    else:
        print("WARNING: Temperature below safety range!")
        print("Heating actuator activated...")

        while temp < 1500:
            temp += 1

        print("Heating mechanism completed.")
        print(f"Temperature stabilized at {temp:.1f}°C")

    print("--------------------------------------")


# Main Program

print("==============================================")
print(" INTELLIGENT METALLURGICAL BLAST FURNACE")
print("       TEMPERATURE CONTROL SYSTEM")
print("==============================================")

print("\nPEAS REPRESENTATION")
print("P - Maintain temperature within safe range")
print("E - Metallurgical Blast Furnace")
print("A - Heating and Cooling Mechanisms")
print("S - Temperature Sensor")

print("\n===== SELECT TEMPERATURE UNIT =====")
print("1. Celsius")
print("2. Fahrenheit")
print("3. Kelvin")
print("4. Exit")


while True:

    try:
        choice = int(input("\nChoose option: "))

        # Celsius
        if choice == 1:

            i_temp = float(input("Enter temperature (°C): "))

            while not 100 <= i_temp <= 2500:
                print("Invalid temperature!")
                i_temp = float(input(
                    "Enter a valid temperature (100-2500 °C): "
                ))

            temp = i_temp
            temp_cont(temp)

        # Fahrenheit
        elif choice == 2:

            i_temp = float(input("Enter temperature (°F): "))

            while not 212 <= i_temp <= 4532:
                print("Invalid temperature!")
                i_temp = float(input(
                    "Enter a valid temperature (212-4532 °F): "
                ))

            # Fahrenheit to Celsius conversion
            temp = (i_temp - 32) * 5 / 9

            temp_cont(temp)

        # Kelvin
        elif choice == 3:

            i_temp = float(input("Enter temperature (K): "))

            while not 373 <= i_temp <= 2773:
                print("Invalid temperature!")
                i_temp = float(input(
                    "Enter a valid temperature (373-2773 K): "
                ))

            # Kelvin to Celsius conversion
            temp = i_temp - 273.15

            temp_cont(temp)

        # Exit
        elif choice == 4:

            print("\nExiting Temperature Control System...")
            print("Thank you!")
            break

        # Invalid menu choice
        else:

            print("Invalid choice! Please select 1, 2, 3 or 4.")

    except ValueError:

        print("Invalid input! Please enter a numeric value.")

```
