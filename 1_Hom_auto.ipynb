```python
# Smart Home Controller Agent
# PEAS:
# Performance Measure: Comfort, safety, and energy efficiency.
# Environment: A home with changing temperature, time, motion, smoke, and rain.
# Actuators: AC, fan, lights, door lock, windows, and alarm.
# Sensors: Temperature sensor, clock, motion sensor, smoke sensor, and rain sensor.

def get_yes_no(message):
    while True:
        answer = input(message + " (yes/no): ").strip().lower()
        if answer in ("yes", "no"):
            return answer
        print("Please enter yes or no.")


def get_temperature():
    while True:
        try:
            value = float(input("Enter room temperature (0-50 C): "))
            if 0 <= value <= 50:
                return value
            print("Temperature must be between 0 and 50.")
        except ValueError:
            print("Please enter a number.")


def get_hour():
    while True:
        try:
            value = int(input("Enter current hour (0-23): "))
            if 0 <= value <= 23:
                return value
            print("Hour must be between 0 and 23.")
        except ValueError:
            print("Please enter a whole number.")


def smart_home_agent(temperature, hour, motion, smoke, rain):
    night = hour >= 18 or hour < 6

    # Smoke emergency has the highest priority.
    if smoke == "yes":
        ac = "OFF"
        fan = "OFF"
        lights = "ON (Emergency)"
        door = "OPEN (Emergency Exit)"
        windows = "OPEN (Smoke Ventilation)"
        alarm = "ON (Fire Alarm)"
    else:
        if temperature >= 26:
            ac = "ON"
            fan = "OFF"
        elif temperature < 20:
            ac = "OFF"
            fan = "OFF"
        else:
            ac = "OFF"
            fan = "ON" if temperature >= 22 else "OFF"

        if night and motion == "yes":
            lights = "ON"
        elif night:
            lights = "DIM"
        else:
            lights = "OFF"

        door = "UNLOCKED" if motion == "yes" else "LOCKED"

        if rain == "yes" or ac == "ON":
            windows = "CLOSED"
        else:
            windows = "OPEN"

        alarm = "ON (Security Alert)" if night and motion == "yes" else "OFF"

    print("\n--- Smart Home Agent Status ---")
    print(f"Temperature: {temperature} C")
    print(f"Time: {hour}:00")
    print(f"AC: {ac}")
    print(f"Fan: {fan}")
    print(f"Lights: {lights}")
    print(f"Door: {door}")
    print(f"Windows: {windows}")
    print(f"Alarm: {alarm}")


# Sensors provide the agent's input.
temperature = get_temperature()
hour = get_hour()
motion = get_yes_no("Is motion detected?")
smoke = get_yes_no("Is smoke detected?")
rain = get_yes_no("Is it raining?")

# The agent processes the sensor input and selects actions.
smart_home_agent(temperature, hour, motion, smoke, rain)
```
