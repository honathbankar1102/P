```python

import heapq

graph = {}
heuristic = {}

n = int(input("Enter the number of Nodes: "))

for i in range(n):
    node = input("Enter node: ")
    h = int(input(f"Enter heuristic value of {node}: "))
    heuristic[node] = h
    graph[node] = []

for node in graph:
    m = int(input(f"How many neighbours of {node}: "))

    for i in range(m):
        child = input("Neighbour: ")
        graph[node].append(child)


def show_lists(open_list, closed):
    open_nodes = [node for _, node in open_list]

    print("OPEN LIST  :", open_nodes)
    print("CLOSED LIST:", list(closed))
    print("-" * 40)


def greedy_best_first_search(start, goal):
    open_list = []
    closed = set()
    parent = {}

    heapq.heappush(open_list, (heuristic[start], start))

    print("\nInitial State:")
    show_lists(open_list, closed)

    while open_list:

        _, current = heapq.heappop(open_list)

        if current in closed:
            continue

        closed.add(current)

        print(f"\nVisiting Node: {current}")

        if current == goal:
            path = []

            while current:
                path.append(current)
                current = parent.get(current)

            path.reverse()

            print("\nGoal Found!")
            print("Path:", " -> ".join(path))
            print("\nFinal State:")
            show_lists(open_list, closed)

            return path

        for neighbour in graph[current]:

            if neighbour not in closed and neighbour not in parent:
                parent[neighbour] = current

                heapq.heappush(
                    open_list,
                    (heuristic[neighbour], neighbour)
                )

        show_lists(open_list, closed)

    return None


start = input("Enter start value: ")
goal = input("Enter goal value: ")

path = greedy_best_first_search(start, goal)

if path:
    print("\nPath:", " -> ".join(path))
else:
    print("\nNo path exists")

```
