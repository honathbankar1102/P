```python

pump_speed = 0

while(True):

    while(True):
        soil_moisture = int(input("Enter soil moisture percentage: "))

        if(soil_moisture < 0 or soil_moisture > 100):
            print("Wrong input, give correct moisture value")
        else:
            break

    water_level = int(input("Enter water tank level percentage: "))

    while(True):
        rain = input("Is it raining? (yes/no): ")

        if(rain == "yes" or rain == "no"):
            break
        else:
            print("Invalid input, give answer only yes or no")

    # Irrigation decision
    if(rain == "yes"):
        print("Rain detected - Turning OFF Water Pump")
        pump_speed = 0

    elif(water_level <= 0):
        print("Water tank is empty - Turning OFF Water Pump")
        pump_speed = 0

    elif(soil_moisture < 20):

        if(pump_speed == 0):
            pump_speed = 1
            print("Soil is very dry - Turning ON Water Pump")

        elif(pump_speed == 3):
            print("Maximum irrigation speed reached")

        else:
            pump_speed += 1
            print("Increasing irrigation speed")

    elif(soil_moisture < 40):

        if(pump_speed == 0):
            pump_speed = 1
            print("Soil is dry - Turning ON Water Pump")

        else:
            print("Maintaining irrigation")

    elif(soil_moisture >= 60):

        if(pump_speed != 0):
            pump_speed = 0
            print("Soil has sufficient water - Turning OFF Water Pump")

        else:
            print("Pump is already OFF")

    else:
        print("Soil moisture is at a suitable level")
        pump_speed = 0

    print("Pump Speed:", pump_speed)

    print("Enter 1 to continue...")
    n = input()

    if(n != '1'):
        print("Turning OFF Smart Irrigation System...")
        break
```
